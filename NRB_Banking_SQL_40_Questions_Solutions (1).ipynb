## Q36 — Recursive CTE

**Task:** Using a recursive CTE, display the employee hierarchy with level and the full chain of command (start from the employee who has no manager).

Write the SQL query in the next cell.

```python
%%sql

WITH org AS (
    SELECT EmployeeID, FullName, Designation, 0 AS Lvl,
           CAST(FullName AS VARCHAR(500)) AS Chain
    FROM Employees
    WHERE ManagerID IS NULL

    UNION ALL

    SELECT e.EmployeeID, e.FullName, e.Designation, o.Lvl + 1,
           CAST(o.Chain + ' > ' + e.FullName AS VARCHAR(500))
    FROM Employees e
    JOIN org o ON e.ManagerID = o.EmployeeID
)
SELECT EmployeeID, Designation, Lvl, Chain
FROM org
ORDER BY Lvl, EmployeeID;
```
